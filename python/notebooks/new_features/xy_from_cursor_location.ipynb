```python
from holoviews import streams

coarse = results.coarsen(x=10, y=10, boundary='trim').mean()
plot = coarse.water_depth.hvplot(x ='y', y = 'x', cmap='Blues', clim=(0,1), aspect='equal')\
    * coarse.dem.hvplot.contour(x ='y', y = 'x',levels = 40, cmap='hot')\
    * coarse.melt.hvplot.contour(x ='y', y = 'x', levels=[0.0, 0.0])\
    * coarse.hvplot.scatter(x = 'y_center_of_mass', y = 'x_center_of_mass', color = 'green', size = 400, marker = '*')
```

```python
pointer = streams.BoundsXY
```

```python
plot = coarse.water_depth.hvplot(x ='y', y = 'x', cmap='Blues', clim=(0,1), aspect='equal')
# Create a PointerXY stream
pointer = streams.PointerXY(x=0, y=0, source=plot)

# Function to print the cursor position
def print_position(x, y):
    print(f'Cursor position: x={x}, y={y}')

# Link the function to the PointerXY stream
pointer.add_subscriber(print_position)

# Display the plot

plot
```
